### Formatting textual columns

```python
movies.genres[0]
```

```python
movies.keywords[0]
```

```python
import ast
# ast.literal_eval safely converts a string that contains a Python literal into the actual Python object. It handles strings, numbers, tuples, lists, dicts, sets, booleans, and None.
def format_genres(recv_string):
    formatted_list = []
    for each_dict in ast.literal_eval(recv_string): #convert string into list here.
        formatted_list.append(each_dict.get('name'))
    return formatted_list
```

```python
movies.genres=movies.genres.apply(format_genres)
movies.keywords=movies.keywords.apply(format_genres)
```

```python
movies.cast[0]
```

```python
def format_cast(recv_string):
    formatted_list = []
    counter=0
    for each_dict in ast.literal_eval(recv_string): 
        if counter<3:
            formatted_list.append(each_dict.get('name'))
        counter+=1
    return formatted_list
```

```python
movies.cast=movies.cast.apply(format_cast)
```

```python
movies.crew[0]
```

```python
def format_crew(recv_list):
    formatted_list = []
    for each_dict in ast.literal_eval(recv_list): 
        if each_dict.get('job')=='Director':
            formatted_list.append(each_dict['name'])
    return formatted_list
```

```python
movies.crew=movies.crew.apply(format_crew)
```

```python
movies.sample(5)
```

```python
movies.overview=movies.overview.apply(lambda x: x.split())
# converted all columns into list
```

```python
movies.sample(3)
```

```python
# we have one problem here there are two entity Sam Worthington and Sam Mendes. There is space between two words so two tags will be genarated like 'sam' and 'worthington' then the model may get confuse which sam are we refering to so the performance may degrade so we must remove space between them.

movies.cast=movies.cast.apply(lambda x: [i.replace(" ","") for i in x])
movies.crew=movies.crew.apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies.genres=movies.genres.apply(lambda x: [i.replace(" ","") for i in x])
movies.keywords=movies.keywords.apply(lambda x: [i.replace(" ","") for i in x])
```
