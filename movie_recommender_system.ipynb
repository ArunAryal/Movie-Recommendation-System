### Formatting textual columns and creating tags

```python
movies.genres[0]
```

```python
import ast
# ast.literal_eval safely converts a string that contains a Python literal into the actual Python object. It handles strings, numbers, tuples, lists, dicts, sets, booleans, and None.
def format_genres(recv_string):
    formatted_list = []
    for each_dict in ast.literal_eval(recv_string): #convert string into list here.
        formatted_list.append(each_dict.get('name'))
    return formatted_list
```

```python
movies.genres=movies.genres.apply(format_genres)
movies.keywords=movies.keywords.apply(format_genres)
```

```python
movies.cast[0]
```

```python
def format_cast(recv_string):
    formatted_list = []
    counter=0
    for each_dict in ast.literal_eval(recv_string): 
        if counter<3:
            formatted_list.append(each_dict.get('name'))
        counter+=1
    return formatted_list
```

```python
movies.cast=movies.cast.apply(format_cast)
```

```python
movies.crew[0]
```

```python
def format_crew(recv_list):
    formatted_list = []
    for each_dict in ast.literal_eval(recv_list): 
        if each_dict.get('job')=='Director':
            formatted_list.append(each_dict['name'])
    return formatted_list
```

```python
movies.crew=movies.crew.apply(format_crew)
```

```python
movies.overview=movies.overview.apply(lambda x: x.split())
# converted all columns into list
```

### Text Vectorization (using Bag of Words technique)



**Vectorization**
Turning text into numbers. Models can't compare words, but they can compare lists of numbers. Each movie's tags become one vector, so we can measure how similar two movies are.

**Bag of Words (BoW)**
One way to vectorize. Build a vocabulary of all words (capped at 5000 here), then count how many times each word appears in each movie's tags. Word order is ignored, as if the words were tossed in a bag.

| Movie   | space | alien | love | war |
|---------|:-----:|:-----:|:----:|:---:|
| Avatar  | 1     | 2     | 1    | 1   |
| Titanic | 0     | 0     | 3    | 0   |

Movies with similar vectors have similar content.

**Stemming**
Cutting words down to their root: `loving`, `loved`, `lovely` → `love`.

**Why stemming is needed**
- Without it, `love`, `loving` and `loved` become 3 separate columns, so similar movies look less similar than they are.
- With it, they merge into 1 column.
- It also shrinks the vocabulary, so the 5000-word limit holds more useful words.

**Order of steps**
Clean the text (lowercase, remove punctuation and stopwords) → stem → vectorize (BoW). BoW counts whatever words it receives, so the text must be cleaned and stemmed first.

#### stemming

```python
# here we can see tags like love,loving,loved which are basically the same so we apply stemmming which does following task
#[love,loving,loved]->[love,love,love]
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
ps.stem('loving')
```

```python
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

def stem_helper(text):
    STOP = ENGLISH_STOP_WORDS | {"arent", "dont", "didnt", "doesnt", "isnt", "wasnt", "cant", "wont"}
    text = re.sub(r"['’]", "", text)       # drop apostrophes: bond’s -> bonds
    text = re.sub(r"[^\w\s]", " ", text)   # other punctuation -> space (keeps é etc.)
    return " ".join(
        ps.stem(w) for w in text.split() if w not in STOP
    )
```

```python
stem_helper('loving loved lovely love')
```

```python
new_df['tags']=new_df['tags'].apply(stem_helper)
```
