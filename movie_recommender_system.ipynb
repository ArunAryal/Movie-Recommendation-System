```python
import numpy as np
import pandas as pd
```

```python
movies=pd.read_csv('dataset/tmdb_5000_movies.csv')
credits=pd.read_csv('dataset/tmdb_5000_credits.csv')
```

```python
movies.sample(3)
```

```python
credits.sample(3)
```

```python
movies=movies.merge(credits,on='title')
movies.sample(2)
```

```python
movies=movies[['movie_id','title','overview','genres','keywords','cast','crew']]
movies.sample(2)
```

```python
movies.shape
```

### handling missing data and duplicate data

```python
movies.isnull().sum()
```

```python
# since only 3 overview are missing and overview is very important feature in our dataset, dropping those rows
movies.dropna(subset=['overview'],inplace=True)
```

```python
movies.isnull().sum()
```

```python
movies.duplicated().sum()
```

### Formatting textual columns

```python
def format_column(list):
    formatted_list=[]
    for dict in list:
        for id,value in dict:
            formatted_list.append(value)
    return formatted_list
```

```python
movies.genres.apply(format_column)
```
